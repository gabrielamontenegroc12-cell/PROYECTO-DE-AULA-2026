```python
from prefect import task, flow
```

```python
@task
def formato_tres_cinco_uno():
    print("tarea 1")
@task
def filtar_solo_usd(formato):
    print("tarea 2")
@task
def solo_usd_rv(filtar):
    print("tarea 3")
@task
def armar_portafolio(rv_usd):
    print("tarea 4")
@task
def compras_ventas(armar):
    print("tarea 5")
```

```python
@flow
def flujo_de_tareas():
    formato = formato_tres_cinco_uno()
    filtrar = filtar_solo_usd(formato)
    rv_usd = solo_usd_rv(filtrar)
    armar = armar_portafolio(rv_usd)
    compras = compras_ventas(armar)
```

```python
flujo_de_tareas()
```
